# tsod: Anomaly Detection for time series data.

![OS](https://img.shields.io/badge/OS-Windows%20%7C%20Linux-blue.png)
![Python version](https://img.shields.io/pypi/pyversions/tsod.svg)
![PyPI version](https://badge.fury.io/py/tsod.svg)

##  **Set up in 5 minutes**

Install **tsod** with [`pip`](https://pypi.org/project/tsod/) and get up
and running in minutes

##  **It’s just Python**

Use familiar Python workflows to integrate anomaly detection into your
models and pipelines

##  **Rule-based detectors**

Choose from detectors like `RangeDetector` and `ConstantValueDetector`
to identify anomalies.
