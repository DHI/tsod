# load

``` python
load(path)
```

Load a saved model from disk saved with `Detector.save`

## Parameters

| Name | Type        | Description                             | Default    |
|------|-------------|-----------------------------------------|------------|
| path | str or Path | File-like object to load detector from. | *required* |

## Returns

| Name | Type     | Description                   |
|------|----------|-------------------------------|
|      | Detector | The loaded detector instance. |
